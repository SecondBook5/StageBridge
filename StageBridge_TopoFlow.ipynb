# StageBridge TopoFlow
**Optimal Transport Flow Matching for Lung Cancer Stage Progression**

---

This notebook demonstrates the complete StageBridge TopoFlow pipeline: from raw cell populations
through HLCA-aligned latent space, entropic OT coupling, and conditional flow-matching training.

### Notebook Scope

| Notebook | Purpose |
|---|---|
| **StageBridge.ipynb** | Data preprocessing: loading raw GEO data, QC, normalization, HVG selection, PCA, Harmony, UMAP, Tangram mapping, niche tokenization. Produces the processed AnnData files that this notebook consumes. |
| **StageBridge_TopoFlow.ipynb** *(this)* | Model pipeline: HLCA latent alignment, Sinkhorn OT coupling, StageBridge flow-matching training, donor-holdout evaluation, artifact generation. Consumes processed AnnData or runs on synthetic data. |

### Pipeline Steps

1. **Configuration** — Load YAML config, resolve paths
2. **Data** — Synthetic AAH/AIS populations (smoke) or processed AnnData (full)
3. **HLCA Latent Alignment** — Project into shared HLCA latent space
4. **Sinkhorn OT Coupling** — Entropic optimal transport between stages
5. **Flow-Matching Training** — StageBridgeModel with donor-holdout CV
6. **Evaluation** — Sinkhorn distance, MMD-RBF, C2ST-AUC, composition JSD
7. **Artifacts** — Config, metrics, report written to disk
